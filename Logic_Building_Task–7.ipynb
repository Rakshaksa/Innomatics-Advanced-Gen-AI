# Problem 1: Smart Light Controller

```python
class Smart_Light:
  def __init__(self,name, status):
    self.name = name
    self.status = status
  def turn_on(self):
    self.status = 'ON'
  def display(self):
    print(self.name, 'is', self.status)
bedroom = Smart_Light('Bedroom Light','OFF')
bedroom.turn_on()
bedroom.display()
```

# Problem 2: Employee ID Card System

```python
class Employee_info:
  def __init__(self, name, ID, department):
    self.name = name
    self.ID = ID
    self.department = department
  def display(self):
    print('Employee Name :', self.name)
    print('Employee ID :', self.ID)
    print('Department :', self.department)
emp1 = Employee_info('Rahul', 'EMP102', 'IT')
emp1.display()
```

# Problem 3: Mobile Contact Record

```python
class Contact_record:
  def __init__(self,name,number):
    self.name = name
    self.number = number
  def display(self):
    print('Contact Saved')
    print('Name :',self.name)
    print('Phone :',self.number)
contact1 = Contact_record('Anita', 9874563210)
contact1.display()
```

# Problem 4: Product Price Tag Generator


```python
class Price_tag:
  def __init__(self, name, price):
    self.name = name
    self.price = price
  def display(self):
    print('Product:',self.name)
    print(f'Price: ₹{self.price}')

product1 = Price_tag('Boat Headphones', 2499)
product1.display()
```

# Problem 5: Movie Rating Display System

```python
class Movie_rating:
  def __init__(self,name,rating):
    self.name = name
    self.rating = rating
  def display(self):
    print('Movie Name:',self.name)
    print(f'Rating: {self.rating} / 5')
movie1 =Movie_rating('Inception', 4.8)
movie2 =Movie_rating('Intersteller', 4.9)

movie1.display()
movie2.display()


```

# Problem 6: Delivery Address Manager

```python
class DeliveryAddress:

    def __init__(self, customer_name, address):
        # Store values inside object
        self.customer_name = customer_name
        self.address = address

    def display(self):
        print("Delivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.address)


# Sample Test
customer1 = DeliveryAddress("Suman", "Hyderabad")
customer1.display()
```
