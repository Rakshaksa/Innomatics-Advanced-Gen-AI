# Problem 1: Social Media: Post Engagement Analyzer

```python
def analyze_engagement(likes_list):
  total = 0
  for like in likes_list:
    total += like
  if total >= 1000:
    Post_Status = 'Viral Post'
  else:
    Post_Status = 'Normal Engagement'

  print('Total likes:', total)
  print('Post Status:', Post_Status)

#test
analyze_engagement([100, 632, 132, 222])
analyze_engagement([145, 332, 212, 310])
```

# Problem 3: Agriculture: Rainfall Adequacy Checker

```python
def rainfall_adequacy_checker(rainfall_list):
  required_level = 110
  total_rainfall = 0
  for rainfall in rainfall_list:
    total_rainfall += rainfall
  avg_rainfall = total_rainfall // len(rainfall_list)
  if avg_rainfall >= required_level:
    Rainfall_Status = 'Adequate Rainfall'
  else:
    Rainfall_Status = 'Inadequate Rainfall'

  print('Average Rainfall:', avg_rainfall)
  print('Rainfall Status:', Rainfall_Status)

#test
rainfall_adequacy_checker([100, 89, 175])
rainfall_adequacy_checker([99, 103, 100])
```

# Problem 4: Social Media: Duplicate Account Detection

```python
def Duplicate_Account_Detection(usernames):
  unique_users = set(usernames)
  if len(usernames) > len(unique_users):
    Duplicate_Account = "Yes"
  else:
    Duplicate_Account = "No"
  print('Duplicate Accounts Found :', Duplicate_Account)

#test
Duplicate_Account_Detection(('Rahul','Aditya','Rahul','Pooja','Abhishek'))
```

# Problem 6: Agriculture: Premium Crop Price Filter

```python
def Premium_Crop_Price_Filter(price):
  premium = []
  regular = []
  for value in price:
    if value > 2000 :
      premium.append(value)
    else:
      regular.append(value)

  print('Premium Crops :', premium)
  print('Regular Crops :', regular)

#test
Premium_Crop_Price_Filter([2500, 1000, 3000, 1500])
```

# Problem 7: System Monitoring: Application Health Checker

```python
def Application_Health_Checker(errors):
  error = len(errors)
  if error == 0 :
    System_Status = 'Healthy'

  elif error <= 5:
    System_Status = 'Minor Issues'

  elif error > 5:
    System_Status = 'Critical Issues'

  else:
    print('No Error')

  print('Error Count:', error)
  print('System Status', System_Status)

#test
Application_Health_Checker(('Database Error', 'Timeout Error', 'API Failure',
                            'Crash Error', 'API Error', 'Memory Leak'))
Application_Health_Checker(('Database Error', 'Timeout Error'))
Application_Health_Checker(())
```

# Problem 8: Banking: Daily Transaction Limit Check

```python
def Transaction_Limit_Checker(amount):
  Daily_limit = 50000
  if amount > Daily_limit :
    Transaction_Status = 'Rejected'
  else:
    Transaction_Status = 'Approved'

  print('Transaction Amount:', amount)
  print('Transaction Status:', Transaction_Status)

#test
Transaction_Limit_Checker(60000)
Transaction_Limit_Checker(30000)
```

# Problem 9: E-Learning: Student Attendance Eligibility System

```python
def Attendance_Eligibility(Attendance_list):
  present_cnt = 0
  for i in Attendance_list:
    if i == 'P':
      present_cnt += 1

  percentage = (present_cnt / len(Attendance_list)) * 100

  if percentage >= 75:
    Exam_Eligibility = 'Eligible'
  else:
    Exam_Eligibility = 'Not Eligible'



  print('Attendance Percentage:', percentage)
  print('Exam Eligibility:', Exam_Eligibility)

#test
Attendance_Eligibility(['P','P','A','P','P'])
Attendance_Eligibility(['P','P','A','P','A'])
```

# Problem 10: Smart Electricity Bill_Amount Analyzer

```python
def Electricity_Bill_Analyzer(units):

  if units <= 100:
    bill = units * 3
  elif units <= 200:
    bill = (100 * 3) + ((units - 100) * 5)
  else:
    bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

  if bill < 500:
    return 'Bill:', bill, 'Usage Status: Low Usage'
  elif bill <= 1500:
    return 'Bill:', bill, 'Usage Status: Moderate Usage'
  else:
    return 'Bill:', bill, 'Usage Status: High Usage'

# test
Electricity_Bill_Analyzer(225)
```
