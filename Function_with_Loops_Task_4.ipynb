# Problem 1: Mobile Recharge Validation System

```python
def validate_recharge(amount):
  valid_plan = [199,299,399,599]
  if amount < 50:
    return False
  if amount in valid_plan:
    return True
  else:
    return False
while True:
  amount = int(input('Enter the Recharge Amount :'))
  if validate_recharge(amount):
    print('Recharge Successful.')
    break
  else:
    print('Invalid recharge amount. Please try again')
```

# Problem 2: Inventory Reorder Alert System


```python
def check_inventory(inventory_data):
  for product, quantity in inventory_data.items():
    if quantity < 15:
      print(f"{product}: Reorder Alert (Stock = {quantity})")
    else:
      print(f"{product}: Stock OK (Stock = {quantity})")

# Sample inventory data
inventory_data = {
    "Rice": 20,
    "Oil": 10,
    "Sugar": 5,
    "Wheat": 18,
    "Salt": 25
}

check_inventory(inventory_data)
```

# Problem 3: Student Result Processing System

```python
def process_result(marks):
  total = 0
  for mark in marks:
    total += mark
  avg = total / len(marks)
  if avg >= 50:
    return "Pass"
  else:
    return "Fail"

# Sample test data
student_marks1 = [60, 55, 40, 70]
student_marks2 = [30, 20, 10, 40]

result1 = process_result(student_marks1)
result2 = process_result(student_marks2)

print("Result for student_marks1:", result1)
print("Result for student_marks2:", result2)
```

# Problem 4: Cab Fare Estimator with Retry Option

```python
def calculate_fare(distance, peak_hour):
    fare = 50 + (distance * 12)
    if peak_hour:
        fare *= 1.25
    return fare

while True:
    try:
        distance = float(input("Enter distance (km): "))
        peak_input = input("Is it peak hour? (y/n): ").lower()

        peak_hour = peak_input == "y"

        total_fare = calculate_fare(distance, peak_hour)
        print(f"Total Fare: ₹{total_fare:.2f}")

        retry = input("Do you want to calculate again? (y/n): ").lower()
        if retry != "y":
            break

    except ValueError:
        print("Please enter valid numeric distance.")
```

# Problem 5: Employee Attendance Eligibility Checker


```python
def check_attendance(attendance_list):
    present_days = 0
    for status in attendance_list:
        if status == "P":
            present_days += 1

    percentage = (present_days / len(attendance_list)) * 100

    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"

# Sample data
attendance = ["P", "A", "P", "P", "A", "P", "P", "P"]

result = check_attendance(attendance)
print("Attendance Status:", result)
```

# Problem 6: Password Strength Checker


```python

def check_password_strength(password):
    has_digit = False
    has_special = False
    special_chars = "@#$"

    if len(password) < 8:
        return "Weak Password"
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch in special_chars:
            has_special = True

    if has_digit and has_special:
        return "Strong Password"
    else:
        return "Weak Password"

pwd = input("Enter password: ")
print(check_password_strength(pwd))
```
