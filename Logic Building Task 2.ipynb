# Problem 1: Unique Words in a Sentence

```python
sentence = "python is easy and python is powerful"
words = sentence.split()
unique_words = set(words)

print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
```

# Problem 2: Highest Salary from Employee Data

```python
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

highest_paid = max(employees, key=employees.get)
print(f"Highest Salary: {highest_paid} : {employees[highest_paid]}")
```

# Problem 3: Find Maximum and Minimum Values

```python
numbers = [45, 22, 89, 10, 66]

maximum = numbers[0]
minimum = numbers[0]
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

print("List:", numbers)
print("Max:", maximum)
print("Min:", minimum)
```

# Problem 4: Count Products Above a Price Threshold

```python
prices = [450, 1200, 899, 1500, 300]
threshold = 1000

count_above_threshold = 0
for price in prices:
    if price > threshold:
        count_above_threshold += 1

print("Products above 1000:", count_above_threshold)
```

# Problem 5: Calculate Attendance Percentage

```python
attendance = ["P", "P", "A", "P", "P"]

present_count = attendance.count("P")
attendance_percentage = (present_count / len(attendance)) * 100

print("Attendance Percentage:", attendance_percentage)
```

# Problem 6: Remove Duplicate Phone Numbers

```python
phone_numbers = [9876543210, 9123456789, 9876543210]
unique_phone_numbers = set(phone_numbers)

print("Unique phone numbers:", unique_phone_numbers)
```

# Problem 7: Count Character Frequency

```python
text = "python"
frequency = {}

for ch in text:
    frequency[ch] = frequency.get(ch, 0) + 1

print(frequency)
```

# Problem 9: Check if a Key Exists in a Dictionary

```python
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

key_to_check = "Ravi"
if key_to_check in employees:
    print("Employee exists")
else:
    print("Employee does not exist")
```

# Problem 10: Calculate Average Marks

```python
marks = [70, 75, 80, 65]
average_marks = sum(marks) / len(marks)

print("Average Marks:", average_marks)
```
